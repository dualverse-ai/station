# Ramsey Numbers for Book Graphs

Problem source: [Epoch AI](https://epoch.ai/frontiermath/open-problems/ramsey-book-graphs)

This notebook gives explicit adjacency-string constructions for the triangular book Ramsey lower-bound task. For each listed integer $n$, the generated graph has exactly $4n - 2$ vertices, contains no $B_{n-1}$, and its complement contains no $B_n$.

The adjacency strings use the column-major upper-triangle convention: for each vertex column $j=1,2,...,N-1$, list the bits for pairs $(0,j),(1,j),...,(j-1,j)$.

## Main Result

The algebraic family proposed here gives a new proof-based family controlled by a different number from the standard Paley family.

The prior Paley family applies under the condition

> all $n$ for which $2n-1$ is a prime power congruent to $1$ mod $4$.

The Seidel/slack family, abbreviated S/K below, is a name used here for the construction in which the Seidel sign matrix $S$ and an auxiliary slack graph $K$ are generated together. It applies under the condition

> all $n$ for which $n-1$ is a prime power congruent to $1$ mod $4$.

These are genuinely different conditions: the older construction is governed by the arithmetic of $2n-1$, while this construction is governed by the arithmetic of $n-1$.

More generally, the Seidel/slack construction only needs $q=n-1$ to admit a conference-type graph with parameters

$$(q, (q-1)/2, (q-5)/4, (q-1)/4).$$

Given such a graph, a fixed four-part/two-extra-vertex rule table constructs a graph on $4n-2$ vertices proving

$$R(B_{n-1}, B_n) > 4n-2.$$

### Comparison under $n <= 100$

The prime-power case of the Seidel/slack family covers the following 15 values with $n <= 100$:

$$n = 6,10,14,18,26,30,38,42,50,54,62,74,82,90,98.$$

There are also two known non-prime-power conference-graph cases under $n <= 100$: $q=45$ from Mathon's PC strongly regular graph and $q=65$ from the Gritsenko graph, giving

$$n = 46, 66.$$

Thus the known conference-graph instances of the Seidel/slack construction cover 17 values under $n <= 100$:

$$n = 6,10,14,18,26,30,38,42,46,50,54,62,66,74,82,90,98.$$

Relative to the [former best results](https://docs.google.com/document/d/1VinXOiMov2v-Y27GTJwoAECywv2a34MH4E8qa712zJY), the Seidel/slack construction adds 6 values not on that list:

$$n = 62, 66, 74, 82, 90, 98.$$

It also gives algebraic, proof-based constructions for 7 values that the former summary had only in the AI-guided heuristic-search category:

$$n = 26, 30, 38, 42, 46, 50, 54.$$

The remaining 4 Seidel/slack values under 100, namely $n=6,10,14,18$, already lie in the classical $n <= 21$ range. Thus, under $n <= 100$, the point of the S/K family is not just that it supplies more bitstrings: it moves several cases from computational-search provenance to an explicit algebraic theorem, and it supplies six additional values beyond the former 71-value list.

The construction is proof-based: once the conference-type graph on $n-1$ vertices is supplied, the graph is generated directly. The code below implements the prime-power Paley source automatically, includes embedded sources for the known non-prime-power cases $q=45$ and $q=65$, and also exposes a lower-level constructor accepting any externally supplied conference-type graph.

## Related Work

We compare our results with Davide Turturean's [progress report](https://docs.google.com/document/d/1VinXOiMov2v-Y27GTJwoAECywv2a34MH4E8qa712zJY/edit?tab=t.0), which lists three algebraic families for this book-Ramsey problem. The first is the previously known standard Paley family, which works when $2n-1$ is a prime power congruent to $1 \pmod 4$. The other two are new uses of existing algebraic constructions from the literature: Szekeres' cyclotomic construction [1], and non-prime-power Paley-type partial difference sets [2]. In all three cases, the construction starts from a known algebraic object and adapts it to the book-Ramsey problem through the known two-block circulant framework.

The algebraic family proposed here is different in kind and is not directly literature-driven. It does not start from the $2n-1$ Paley condition, the Szekeres $4n-1$ prime condition, or the known Paley-type partial-difference-set conditions. Instead, it starts from a conference-type graph on $n-1$ vertices. From that graph, a four-part/two-extra-vertex transformation constructs a graph on $4n-2$ vertices, while the same rule table also produces a Seidel/slack certificate proving the two book-avoidance inequalities.

To the best of our knowledge, this four-part lift from a conference graph on $n-1$ vertices to a book-Ramsey construction on $4n-2$ vertices has not appeared in the literature. Thus the result is not merely another use of an existing Paley/Szekeres/PDS family; it gives a new algebraic mechanism for producing explicit verifiable constructions.

## Method And Proof

The method starts from a conference-type graph $R$ on a coordinate set $X$ of size $q=n-1$. It builds four indexed parts $BB, BR, RB, RR$, each a copy of $X$, and adds two additional vertices $u,v$. A fixed rule table assigns each pair a Seidel sign $S_{ij}=+1$ for an edge of the constructed graph and $S_{ij}=-1$ for a non-edge, with $S_{ii}=0$. The same table also defines an auxiliary $0/1$ matrix $K$; the proof verifies that $K_{ij}=1$ is exactly the case where the signed square has value $-4$.

Let $R(x,y)$ mean that $x$ and $y$ are adjacent in the input graph $R$. The table used by the constructor is:

| pair type | coordinate case | $S$ | $K$ |
|---|---:|---:|---:|
| BB,BB | R / not R | +1 / -1 | 0 / 1 |
| BR,BR | R / not R | -1 / +1 | 0 / 1 |
| RB,RB | R / not R | -1 / +1 | 0 / 1 |
| RR,RR | R / not R | +1 / -1 | 0 / 1 |
| BB,BR | diagonal / R / not R | +1 / -1 / +1 | 0 / 0 / 0 |
| BB,RB | diagonal / R / not R | +1 / +1 / -1 | 0 / 0 / 0 |
| BB,RR | diagonal / R / not R | -1 / -1 / +1 | 1 / 1 / 0 |
| BR,RB | diagonal / R / not R | -1 / -1 / +1 | 0 / 0 / 1 |
| BR,RR | diagonal / R / not R | -1 / -1 / +1 | 0 / 0 / 0 |
| RB,RR | diagonal / R / not R | -1 / +1 / -1 | 0 / 0 / 0 |
| u with BB,BR,RB,RR | all coordinates | -1,-1,+1,+1 | 0,1,0,0 |
| v with BB,BR,RB,RR | all coordinates | -1,+1,-1,+1 | 0,0,1,0 |
| u,v | endpoint pair | -1 | 0 |

**Theorem.** Let $R$ be a simple graph on $q$ vertices with degree $k=(q-1)/2$, adjacent common-neighbor count $\lambda=(q-5)/4$, and non-adjacent common-neighbor count $\mu=(q-1)/4$. Applying the table above gives a graph $G$ on $N=4q+2=4n-2$ vertices such that $G$ contains no $B_{n-1}$ and the complement of $G$ contains no $B_n$.

**Proof.** The graph $R$ has exactly the two intersection numbers needed for every coordinate pair: adjacent pairs have $\lambda$ common neighbors, non-adjacent pairs have $\mu$ common neighbors, and each coordinate has degree $k$. Therefore every two-step signed count in the four-part table is determined by only the part pair and by whether the coordinates are equal, adjacent in $R$, or non-adjacent in $R$.

There are six row-sum types: the four indexed parts and the two additional vertices. Substituting $k=(q-1)/2$ into the table gives Seidel row sum $-1$ in each case. For example, the row of a $BB$ vertex receives same-part contribution $k-(q-1-k)$, cross-part contributions $1-k+(q-1-k)$, $1+k-(q-1-k)$, and $-1-k+(q-1-k)$, and contribution $-1-1$ from $u$ and $v$, for total $-1$. The other five row types are the same finite table calculation.

For off-diagonal entries, $(S^2)_{ij}=\sum_t S_{it}S_{tj}$. Each pair type is again one of finitely many cases: same part with $R$ or not $R$; distinct parts with equal, $R$-related, or non-$R$-related coordinates; one additional vertex and one part vertex; or the pair $u,v$. Counting intermediate coordinates with $k,\lambda,\mu$ and then substituting the conference parameters gives

$$(S^2)_{ij}=0 \quad\text{when } K_{ij}=0, \qquad (S^2)_{ij}=-4 \quad\text{when } K_{ij}=1.$$

This is an exhaustive finite case check of the displayed rule table; no search or numerical optimization enters the argument.

It remains to translate this signed certificate into book bounds. Since every Seidel row sum is $-1$, every vertex has degree $(N-2)/2=2q$. Consider an edge $ij$ of $G$. If $r$ is the number of common neighbors of $i$ and $j$ in $G$, then a direct count using the two degree equations gives

$$(S^2)_{ij}=4(r-(q-1)).$$

Because the construction gives $(S^2)_{ij}\le 0$, every edge has at most $q-1=n-2$ common neighbors, so $G$ has no $B_{n-1}$.

Now consider a non-edge $ij$ of $G$, equivalently an edge in the complement. If $b$ is the number of common neighbors of $i$ and $j$ in the complement, the same degree count gives

$$(S^2)_{ij}=4(b-q).$$

Again $(S^2)_{ij}\le 0$, so $b\le q=n-1$, and the complement has no $B_n$. This proves the theorem.

```python
#@title Construction functions
from __future__ import annotations

import hashlib
import itertools
from dataclasses import dataclass
from math import isqrt
from typing import Iterable


CHAMBERS = ('BB', 'BR', 'RB', 'RR')
CHAMBER_INDEX = {name: i for i, name in enumerate(CHAMBERS)}

CHAMBER_RULES = {
    'BB|BB': {'R': (1, 0), 'not_R': (-1, 1)},
    'BB|BR': {'diag': (1, 0), 'R': (-1, 0), 'not_R': (1, 0)},
    'BB|RB': {'diag': (1, 0), 'R': (1, 0), 'not_R': (-1, 0)},
    'BB|RR': {'diag': (-1, 1), 'R': (-1, 1), 'not_R': (1, 0)},
    'BR|BR': {'R': (-1, 0), 'not_R': (1, 1)},
    'BR|RB': {'diag': (-1, 0), 'R': (-1, 0), 'not_R': (1, 1)},
    'BR|RR': {'diag': (-1, 0), 'R': (-1, 0), 'not_R': (1, 0)},
    'RB|RB': {'R': (-1, 0), 'not_R': (1, 1)},
    'RB|RR': {'diag': (-1, 0), 'R': (1, 0), 'not_R': (-1, 0)},
    'RR|RR': {'R': (1, 0), 'not_R': (-1, 1)},
}

ENDPOINT_RULES = {
    'u': {'BB': (-1, 0), 'BR': (-1, 1), 'RB': (1, 0), 'RR': (1, 0)},
    'v': {'BB': (-1, 0), 'BR': (1, 0), 'RB': (-1, 1), 'RR': (1, 0)},
    'u|v': (-1, 0),
}


def is_prime(value: int) -> bool:
    if value < 2:
        return False
    if value % 2 == 0:
        return value == 2
    for divisor in range(3, isqrt(value) + 1, 2):
        if value % divisor == 0:
            return False
    return True


def prime_power_decomposition(value: int) -> tuple[int, int] | None:
    if value < 2:
        return None
    if is_prime(value):
        return value, 1
    for p in range(2, isqrt(value) + 1):
        if value % p != 0 or not is_prime(p):
            continue
        remaining = value
        exponent = 0
        while remaining % p == 0:
            remaining //= p
            exponent += 1
        if remaining == 1:
            return p, exponent
    return None


def _trim(poly: list[int]) -> list[int]:
    while len(poly) > 1 and poly[-1] == 0:
        poly.pop()
    return poly


def _poly_mod(poly: list[int], divisor: list[int], p: int) -> list[int]:
    out = [coeff % p for coeff in poly]
    divisor = _trim([coeff % p for coeff in divisor[:]])
    if divisor == [0]:
        raise ZeroDivisionError('zero polynomial divisor')
    while len(out) >= len(divisor) and out != [0]:
        coeff = out[-1] % p
        offset = len(out) - len(divisor)
        if coeff:
            for i, div_coeff in enumerate(divisor):
                out[offset + i] = (out[offset + i] - coeff * div_coeff) % p
        _trim(out)
    return out


def _is_irreducible_monic(poly: list[int], p: int) -> bool:
    degree = len(poly) - 1
    if degree <= 0 or poly[-1] % p != 1 or poly[0] % p == 0:
        return False
    for divisor_degree in range(1, degree // 2 + 1):
        for coeffs in itertools.product(range(p), repeat=divisor_degree):
            divisor = list(coeffs) + [1]
            if divisor[0] % p == 0:
                continue
            if _poly_mod(poly, divisor, p) == [0]:
                return False
    return True


def find_irreducible_polynomial(p: int, degree: int) -> tuple[int, ...]:
    if degree == 1:
        return (0, 1)
    for coeffs in itertools.product(range(p), repeat=degree):
        candidate = list(coeffs) + [1]
        if _is_irreducible_monic(candidate, p):
            return tuple(candidate)
    raise RuntimeError(f'No irreducible polynomial found over F_{p} of degree {degree}')


@dataclass(frozen=True)
class FiniteField:
    p: int
    degree: int
    modulus: tuple[int, ...]

    @classmethod
    def of_size(cls, q: int) -> FiniteField:
        decomposition = prime_power_decomposition(q)
        if decomposition is None:
            raise ValueError(f'{q} is not a prime power')
        p, degree = decomposition
        return cls(p, degree, find_irreducible_polynomial(p, degree))

    @property
    def size(self) -> int:
        return self.p ** self.degree

    def coeffs(self, value: int) -> tuple[int, ...]:
        out = []
        x = value
        for _ in range(self.degree):
            out.append(x % self.p)
            x //= self.p
        return tuple(out)

    def encode(self, coeffs: Iterable[int]) -> int:
        value = 0
        factor = 1
        for coeff in coeffs:
            value += (coeff % self.p) * factor
            factor *= self.p
        return value

    def sub(self, a: int, b: int) -> int:
        if self.degree == 1:
            return (a - b) % self.p
        a_coeffs = self.coeffs(a)
        b_coeffs = self.coeffs(b)
        return self.encode((a_coeffs[i] - b_coeffs[i]) % self.p for i in range(self.degree))

    def mul(self, a: int, b: int) -> int:
        if self.degree == 1:
            return (a * b) % self.p
        a_coeffs = self.coeffs(a)
        b_coeffs = self.coeffs(b)
        product = [0] * (2 * self.degree - 1)
        for i, ai in enumerate(a_coeffs):
            for j, bj in enumerate(b_coeffs):
                product[i + j] = (product[i + j] + ai * bj) % self.p
        for deg in range(len(product) - 1, self.degree - 1, -1):
            coeff = product[deg] % self.p
            if coeff:
                offset = deg - self.degree
                for r in range(self.degree):
                    product[offset + r] = (product[offset + r] - coeff * self.modulus[r]) % self.p
        return self.encode(product[: self.degree])


def paley_conference_relation(q: int) -> tuple[int, ...]:
    if q % 4 != 1:
        raise ValueError(f'q must be congruent to 1 mod 4, got {q}')
    field = FiniteField.of_size(q)
    squares = {field.mul(x, x) for x in range(1, q)}
    masks = [0] * q
    for j in range(1, q):
        for i in range(j):
            if field.sub(j, i) in squares:
                masks[i] |= 1 << j
                masks[j] |= 1 << i
    return tuple(masks)


KNOWN_NON_PRIME_POWER_CONFERENCE_MASKS = {
    45: (
        11131596448600, 18340564747504, 5781006653864, 5840177513669,
        11125000575363, 18287932351302, 18347084027947, 5833562661406,
        11072348931125, 34693475381409, 31326175093002, 4388151316564,
        34683631536724, 31316388284065, 4378388237578, 34664046679818,
        31296827178068, 4358746434209, 30135862510584, 30112347198919,
        30118032484415, 25095814031864, 25101501023175, 25119336043583,
        15068367295992, 15086214752711, 15062732718655, 18806630511030,
        6766790938550, 9677657112502, 6760971891053, 9672711442285,
        18804241891181, 9626568093915, 18764465934043, 6741664497371,
        23661241396497, 16528495832162, 29148349164684, 13549143513698,
        26616166013580, 22438268033809, 2990771648140, 2072977979153,
        3663757948514,
    ),
    65: (
        8589934590, 32702208444638191557, 30240416579365388195,
        20119351150514634513, 10290220086755262089, 4376147643626814533,
        6583361713055332899, 26332734368684175639, 17503614208183756943,
        31544101928124392543, 33120827962859045439, 15496646957681906047,
        21803832571597097215, 13572891922385486845, 24948561726703805947,
        26007270602816950257, 17398079551283304425, 33510637918320685201,
        30594022942441138441, 23371243961714189349, 11686470596818879043,
        9888598950114168983, 19661794841760401679, 2555072563577889373,
        4966038187432494139, 19865841719344269553, 10216912409067983209,
        5666952402643198917, 3981910870427452835, 22844611630075252369,
        15749152612347483913, 17592446566152420389, 26103834759602264643,
        26990002577574900504, 17228415062928870552, 34180774603065470052,
        32010742718371254882, 26584191825197879700, 16822637425894664586,
        32559352333827229268, 30389189026606687786, 19558633734357555372,
        10877980496814217554, 6753964032542401228, 4313717273977840946,
        17794172952268475056, 26474863423952622920, 31976797666115204294,
        34417194546651613990, 4161623950616255036, 6458328401007262810,
        16645405077811685748, 25832715478822121706, 29545143459604710828,
        29688563174533080018, 7500672209221097270, 8074914292675317454,
        29991992814033984860, 32306975545370360506, 9288632851256112368,
        18548000551707306856, 259785424893227940, 404594920612679106,
        1038543556688949014, 1617288949704611470,
    ),
}

KNOWN_NON_PRIME_POWER_CONFERENCE_SOURCES = {
    45: 'Mathon PC strongly regular graph with parameters (45,22,10,11)',
    65: 'Gritsenko strongly regular graph with parameters (65,32,15,16)',
}


def known_conference_relation(q: int) -> tuple[int, ...]:
    if q in KNOWN_NON_PRIME_POWER_CONFERENCE_MASKS:
        return KNOWN_NON_PRIME_POWER_CONFERENCE_MASKS[q]
    if q % 4 == 1 and prime_power_decomposition(q) is not None:
        return paley_conference_relation(q)
    raise ValueError(
        'automatic construction requires n - 1 to be either a prime power congruent to 1 mod 4 '
        'or one of the embedded non-prime-power conference cases q=45,65'
    )


def conference_relation_source(q: int) -> str:
    if q in KNOWN_NON_PRIME_POWER_CONFERENCE_SOURCES:
        return KNOWN_NON_PRIME_POWER_CONFERENCE_SOURCES[q]
    if q % 4 == 1 and prime_power_decomposition(q) is not None:
        return f'Paley conference graph over GF({q})'
    return 'not available in this notebook'


def relation_masks_from_edges(vertex_count: int, edges: Iterable[tuple[int, int]]) -> tuple[int, ...]:
    masks = [0] * vertex_count
    for a, b in edges:
        if a == b:
            raise ValueError('The source relation must be loopless')
        masks[a] |= 1 << b
        masks[b] |= 1 << a
    return tuple(masks)


def conference_relation_stats(r_masks: tuple[int, ...]) -> dict[str, object]:
    q = len(r_masks)
    expected_degree = (q - 1) // 2
    expected_lambda = (q - 5) // 4
    expected_mu = (q - 1) // 4
    loop_count = 0
    symmetry_mismatches = 0
    degrees = []
    adjacent_common = []
    nonadjacent_common = []
    for i, mask_i in enumerate(r_masks):
        loop_count += (mask_i >> i) & 1
        degrees.append((mask_i & ~(1 << i)).bit_count())
        for j in range(i + 1, q):
            symmetry_mismatches += int(((r_masks[i] >> j) & 1) != ((r_masks[j] >> i) & 1))
    for j in range(1, q):
        for i in range(j):
            common = (r_masks[i] & r_masks[j]).bit_count()
            if (r_masks[i] >> j) & 1:
                adjacent_common.append(common)
            else:
                nonadjacent_common.append(common)
    ok = (
        q % 4 == 1
        and loop_count == 0
        and symmetry_mismatches == 0
        and set(degrees) == {expected_degree}
        and set(adjacent_common) == {expected_lambda}
        and set(nonadjacent_common) == {expected_mu}
    )
    return {
        'ok': ok,
        'q': q,
        'degree_values': sorted(set(degrees)),
        'adjacent_common_values': sorted(set(adjacent_common)),
        'nonadjacent_common_values': sorted(set(nonadjacent_common)),
        'expected': (expected_degree, expected_lambda, expected_mu),
        'loop_count': loop_count,
        'symmetry_mismatches': symmetry_mismatches,
    }


def _decode_vertex(index: int, q: int) -> tuple[str, str | None, int | None]:
    if index < 4 * q:
        return 'chamber', CHAMBERS[index // q], index % q
    return 'endpoint', 'u' if index == 4 * q else 'v', None


def sk_rule_for_pair(i: int, j: int, q: int, r_masks: tuple[int, ...]) -> tuple[int, int]:
    type_i, label_i, coord_i = _decode_vertex(i, q)
    type_j, label_j, coord_j = _decode_vertex(j, q)
    if type_i == 'endpoint' and type_j == 'endpoint':
        return ENDPOINT_RULES['u|v']
    if type_i == 'endpoint' or type_j == 'endpoint':
        endpoint = label_i if type_i == 'endpoint' else label_j
        chamber = label_j if type_i == 'endpoint' else label_i
        return ENDPOINT_RULES[endpoint][chamber]
    if CHAMBER_INDEX[label_i] > CHAMBER_INDEX[label_j]:
        label_i, label_j = label_j, label_i
        coord_i, coord_j = coord_j, coord_i
    key = f'{label_i}|{label_j}'
    if coord_i == coord_j:
        status = 'diag'
    else:
        status = 'R' if (r_masks[coord_i] >> coord_j) & 1 else 'not_R'
    return CHAMBER_RULES[key][status]


def adjacency_string_from_masks(masks: tuple[int, ...] | list[int]) -> str:
    return ''.join('1' if (int(masks[j]) >> i) & 1 else '0' for j in range(1, len(masks)) for i in range(j))


def build_sk_adjacency_from_relation(r_masks: tuple[int, ...], *, strict: bool = True) -> str:
    check = conference_relation_stats(r_masks)
    if strict and not check['ok']:
        raise ValueError(f'R does not have the required conference parameters: {check}')
    q = len(r_masks)
    vertex_count = 4 * q + 2
    masks = [0] * vertex_count
    for j in range(1, vertex_count):
        for i in range(j):
            sign, _slack = sk_rule_for_pair(i, j, q, r_masks)
            if sign == 1:
                masks[i] |= 1 << j
                masks[j] |= 1 << i
    return adjacency_string_from_masks(masks)


def construct_book_ramsey_witness(n: int) -> str:
    q = n - 1
    if q <= 1 or q % 4 != 1:
        raise ValueError('automatic construction requires n - 1 > 1 and congruent to 1 mod 4')
    return build_sk_adjacency_from_relation(known_conference_relation(q))


def construct_book_ramsey_witness_metadata(n: int) -> dict[str, object]:
    adjacency = construct_book_ramsey_witness(n)
    q = n - 1
    return {
        'n': n,
        'q': q,
        'vertices': 4 * n - 2,
        'q_prime_power': prime_power_decomposition(q),
        'conference_source': conference_relation_source(q),
        'bit_length': len(adjacency),
        'sha256': hashlib.sha256(adjacency.encode()).hexdigest(),
    }


def accepted_n_values(max_n: int) -> list[int]:
    return [
        n for n in range(1, max_n + 1)
        if n - 1 > 1
        and (n - 1) % 4 == 1
        and (
            prime_power_decomposition(n - 1) is not None
            or (n - 1) in KNOWN_NON_PRIME_POWER_CONFERENCE_MASKS
        )
    ]
```

```python
#@title Verification functions
from __future__ import annotations

import hashlib


def parse_adjacency_string(text: str, expected_vertices: int) -> tuple[int, ...]:
    '''Parse a column-major upper-triangle adjacency bitstring into integer masks.'''
    bits = ''.join(str(text).split())
    if any(ch not in '01' for ch in bits):
        raise AssertionError('Adjacency string must contain only 0/1 characters.')

    expected_len = expected_vertices * (expected_vertices - 1) // 2
    if len(bits) != expected_len:
        raise AssertionError(
            f'Expected C({expected_vertices}, 2) = {expected_len} bits, got {len(bits)}.'
        )

    masks = [0] * expected_vertices
    index = 0
    for j in range(1, expected_vertices):
        bit_j = 1 << j
        for i in range(j):
            if bits[index] == '1':
                masks[i] |= bit_j
                masks[j] |= 1 << i
            index += 1
    return tuple(masks)


def triangular_book_stats(adjacency: str, n: int) -> dict[str, int | float | bool | str]:
    '''Return exact book-obstruction statistics for one triangular-book witness.'''
    vertices = 4 * n - 2
    masks = parse_adjacency_string(adjacency, vertices)
    all_mask = (1 << vertices) - 1
    complement_masks = tuple((all_mask ^ (1 << u)) ^ masks[u] for u in range(vertices))

    red_cap = n - 2
    blue_cap = n - 1
    red_max = 0
    blue_max = 0
    red_bad_pairs = 0
    blue_bad_pairs = 0
    edge_count = 0

    for j in range(1, vertices):
        mask_j = masks[j]
        comp_j = complement_masks[j]
        for i in range(j):
            if (mask_j >> i) & 1:
                edge_count += 1
                common = (masks[i] & mask_j).bit_count()
                red_max = max(red_max, common)
                red_bad_pairs += int(common > red_cap)
            else:
                common = (complement_masks[i] & comp_j).bit_count()
                blue_max = max(blue_max, common)
                blue_bad_pairs += int(common > blue_cap)

    total_pairs = vertices * (vertices - 1) // 2
    bits = ''.join(str(adjacency).split())
    return {
        'n': n,
        'vertices': vertices,
        'bit_length': len(bits),
        'edge_count': edge_count,
        'density': edge_count / total_pairs,
        'red_max': red_max,
        'red_cap': red_cap,
        'blue_max': blue_max,
        'blue_cap': blue_cap,
        'red_bad_pairs': red_bad_pairs,
        'blue_bad_pairs': blue_bad_pairs,
        'valid': red_max <= red_cap and blue_max <= blue_cap,
        'sha256': hashlib.sha256(bits.encode()).hexdigest(),
    }


def verify_triangular_book_witness(adjacency: str, n: int) -> dict[str, int | float | bool | str]:
    '''Verify that G avoids B_{n-1} and complement(G) avoids B_n.'''
    stats = triangular_book_stats(adjacency, n)
    red_max = stats['red_max']
    red_cap = stats['red_cap']
    blue_max = stats['blue_max']
    blue_cap = stats['blue_cap']
    assert red_max <= red_cap, (
        f'n={n}: G contains a red obstruction, red_max={red_max} > {red_cap}'
    )
    assert blue_max <= blue_cap, (
        f'n={n}: complement(G) contains a blue obstruction, blue_max={blue_max} > {blue_cap}'
    )
    vertices = stats['vertices']
    edge_count = stats['edge_count']
    print(
        f'Verified n={n}: |V|={vertices}, edges={edge_count}, '
        f'red_max={red_max}/{red_cap}, '
        f'blue_max={blue_max}/{blue_cap}.'
    )
    return stats


def verify_witness_collection(witnesses: dict[int, str]) -> dict[int, dict[str, int | float | bool | str]]:
    '''Verify all witnesses in a dictionary keyed by n.'''
    all_stats = {}
    for n in sorted(witnesses):
        all_stats[n] = verify_triangular_book_witness(witnesses[n], n)
    print(f'Verified {len(all_stats)} triangular-book Ramsey witnesses.')
    return all_stats
```
