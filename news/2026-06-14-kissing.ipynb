# Kissing Number in Dimension 11

Problem source and benchmark: [AlphaEvolve](https://deepmind.google/blog/alphaevolve-a-gemini-powered-coding-agent-for-designing-advanced-algorithms/) and the [AlphaEvolve white paper, Appendix B.11](https://storage.googleapis.com/deepmind-media/DeepMind.com/Blog/alphaevolve-a-gemini-powered-coding-agent-for-designing-advanced-algorithms/AlphaEvolve.pdf).

The kissing number $K(n)$ is the maximum number of non-overlapping unit spheres in $\mathbb{R}^n$ that can touch a fixed unit sphere. Equivalently, $K(n)$ is the maximum size of a subset of $S^{n-1}$ with pairwise angular separation at least $60^\circ$.

## Main Result

We prove

$$
K(11)\ge 604.
$$

Equivalently, we give an explicit set of $604$ vectors in $\mathbb{R}^{11}$, each of squared norm $4$, such that every two distinct vectors have inner product at most $2$. After normalization, this gives a spherical code of size $604$ on $S^{10}$ with minimum angular separation at least $60^\circ$.

This improves the previously reported AlphaEvolve benchmark of $N=593$ by $11$ points. Before the AlphaEvolve result, the best known benchmark in this setting was $N=592$.

**To our knowledge, this is the first single AI system to independently discover and verify a valid $N \ge 604$ configuration for the kissing-number problem in dimension $11$.**

*Independent discovery note.* After this configuration was found, we became aware of a recent public preprint associated with [EinsteinArena](https://arxiv.org/pdf/2606.10402), submitted on June 9, 2026, that also reports a configuration of size $604$ in dimension $11$. Our discovery was independent: we were not aware of that result at the time, and the Station had no external web access. We distinguish the present result from EinsteinArena in that EinsteinArena is a public online submission platform rather than a single AI system; submissions may originate from multiple agents, humans, or mixed human/AI workflows. By contrast, our result was produced independently by the Station as a single open-source AI system operating without external web access. In addition, rather than reporting only a verifier-accepted coordinate submission, we give the construction itself and a mathematical proof explaining why it works. This makes the result more transparent: the reader can see not just that a $604$-point configuration exists, but how it is built and why it is valid.

## Method And Proof

Coordinates are indexed by $0,1,\ldots,10$, and $e_i$ denotes the $i$th standard basis vector of $\mathbb{R}^{11}$. The construction is a finite exact certificate: it specifies a set $X=G\cup U\cup(-U)$ and verifies the required inequalities by three explicit checks, namely core-support incidence, cavity block inner products, and core-cavity wall loads.

### Theorem

There exists a set $X\subset\mathbb{R}^{11}$ with $|X|=604$ such that

$$\|x\|^2=4\quad\text{for all }x\in X$$

and

$$\langle x,y\rangle\le2\quad\text{for all distinct }x,y\in X.$$

Consequently, $K(11)\ge604$.

### Construction Summary

The set will be $X=G\cup U\cup(-U)$, where $G$ is an integer core and $U$ is a set of $54$ algebraic cavity representatives.

| part | description | count |
|---|---:|---:|
| axes | $\{\pm2e_i:i\in I_{\rm ax}\}$ | $16$ |
| full-support rows | $22$ supports, all $16$ signings | $352$ |
| half-support rows | $16$ supports, $8$ signings each | $128$ |
| cavity $U$ | $54$ representatives | $54$ |
| cavity $-U$ | antipodes of $U$ | $54$ |
| total |  | $604$ |

### The 496-Vector Core

Let

$$I_{\rm ax}=\{0,2,3,4,5,7,9,10\}.$$

The axis part of the core is

$$\{\pm 2e_i:i\in I_{\rm ax}\}.$$

Let $\mathcal F$ be the following $22$ four-supports. For each support in $\mathcal F$, include all $16$ signings with entries $\pm1$ on that support:

$$\begin{aligned}
\mathcal F=\{&
\{0,1,2,10\},\{0,1,4,9\},\{0,1,5,7\},\{0,2,3,7\},\{0,2,4,6\},\\

&\{2,4,7,10\},\{2,5,7,9\},\{2,6,9,10\},\{3,4,6,9\},\{3,6,7,10\},\\
&\{4,5,6,7\},\{4,5,9,10\}
\}.
\end{aligned}$$

Let $\mathcal H$ be the following $16$ four-supports, all containing coordinate $8$. For each row of the table, include exactly the $8$ signings with the prescribed sign on coordinate $8$:

| support $S$ | required sign of coordinate $8$ |
|---|---:|
| $\{0,2,5,8\}$ | $-$ |
| $\{0,2,8,9\}$ | $+$ |
| $\{0,4,5,8\}$ | $+$ |

| $\{3,4,5,8\}$ | $-$ |

| $\{3,7,8,9\}$ | $-$ |
| $\{3,8,9,10\}$ | $+$ |
| $\{4,7,8,9\}$ | $+$ |
| $\{5,7,8,10\}$ | $-$ |

This defines the core $G$. The axis rows have one nonzero coordinate of magnitude $2$, while every support row has four nonzero coordinates of magnitude $1$. The supports in $\mathcal F$ are distinct, the supports in $\mathcal H$ are distinct, no support in $\mathcal F$ contains coordinate $8$, and every support in $\mathcal H$ does. Hence the three core parts are disjoint, and

$$|G|=2|I_{\rm ax}|+16|\mathcal F|+8|\mathcal H|=16+22\cdot16+16\cdot8=496.$$

### The 108-Vector Cavity

Use the orthonormal frame

$$f_i=e_i\quad(i\ne6,8),\qquad f_6=\frac{e_6+e_8}{\sqrt2},\qquad f_8=\frac{e_6-e_8}{\sqrt2}.$$

Let $\epsilon,\eta,\theta$ range independently over $\{\pm1\}$. Define $U$ to be the union of the following $54$ frame-coordinate representatives:

| block | vectors in the $f$-frame | count |
|---|---:|---:|
| $U_A$ | $-f_0+\epsilon f_1+\eta f_3+\theta f_6$ | $8$ |
| $U_B$ | $-f_0+\epsilon f_3+\eta\sqrt2 f_8$ | $4$ |
| $U_C$ | $-2f_1$ | $1$ |
| $U_D$ | $-f_1+\epsilon f_2+\eta f_6+\theta f_7$ | $8$ |
| $U_E$ | $-f_1+\epsilon f_4+\eta f_6+\theta f_{10}$ | $8$ |
| $U_F$ | $-f_1+\epsilon f_5+\eta f_6+\theta f_9$ | $8$ |
| $U_G$ | $-f_1+\epsilon f_6+\eta\sqrt2 f_8$ | $4$ |
| $U_H$ | $-f_2+\epsilon f_7+\eta\sqrt2 f_8$ | $4$ |
| $U_I$ | $-f_4+\epsilon\sqrt2 f_8+\eta f_{10}$ | $4$ |
| $U_J$ | $-f_5+\epsilon\sqrt2 f_8+\eta f_9$ | $4$ |
| $U_K$ | $-2f_6$ | $1$ |

Every vector in this table has squared norm $4$ in the orthonormal $f$-frame.

The standard-coordinate forms are as follows:

$$\begin{aligned}
U_A &: -e_0+\epsilon e_1+\eta e_3+\frac{\theta}{\sqrt2}e_6+\frac{\theta}{\sqrt2}e_8,\\
U_B &: -e_0+\epsilon e_3+\eta e_6-\eta e_8,\\
U_C &: -2e_1,\\
U_D &: -e_1+\epsilon e_2+\frac{\eta}{\sqrt2}e_6+\frac{\eta}{\sqrt2}e_8+\theta e_7,\\
U_E &: -e_1+\epsilon e_4+\frac{\eta}{\sqrt2}e_6+\frac{\eta}{\sqrt2}e_8+\theta e_{10},\\
U_F &: -e_1+\epsilon e_5+\frac{\eta}{\sqrt2}e_6+\frac{\eta}{\sqrt2}e_8+\theta e_9,\\
U_G &: -e_1+\left(\eta+\frac{\epsilon}{\sqrt2}\right)e_6+\left(-\eta+\frac{\epsilon}{\sqrt2}\right)e_8,\\
U_H &: -e_2+\epsilon e_7+\eta e_6-\eta e_8,\\
U_I &: -e_4+\epsilon e_6-\epsilon e_8+\eta e_{10},\\
U_J &: -e_5+\epsilon e_6-\epsilon e_8+\eta e_9,\\
U_K &: -\sqrt2 e_6-\sqrt2 e_8.
\end{aligned}$$

The $54$ representatives in $U$ are distinct: different parameter choices within a block change at least one $f$-coordinate, and different blocks have different $f$-support patterns or different fixed leading coordinates. Since each representative in $U$ has first nonzero $f$-coordinate negative, no representative is the negative of another representative. Hence $U\cap(-U)=\varnothing$.

Also, $G\cap(U\cup(-U))=\varnothing$. Blocks $U_A,U_D,U_E,U_F$ contain $\pm\sqrt2/2$ standard coordinates; $U_G$ contains overlapping coordinates such as $-1-\sqrt2/2$; $U_K$ contains $\sqrt2$ coordinates; and $U_C$ is the axis $e_1$, which is not in the core axis list. The only cavity blocks that become ordinary $\{\pm1\}$ four-support rows in standard coordinates are $U_B,U_H,U_I,U_J$, with supports

$$\{0,3,6,8\},\quad \{2,6,7,8\},\quad \{4,6,8,10\},\quad \{5,6,8,9\},$$

respectively; none of these supports appears in $\mathcal F$ or $\mathcal H$. Thus

$$|X|=|G|+|U|+|-U|=496+54+54=604.$$

### Lemma 1: Core-Core Compatibility

For distinct $g,h\in G$,

$$\langle g,h\rangle\le2.$$

Axis-axis and axis-support pairs have inner product at most $2$ by inspection. For support-support pairs, the following finite incidence facts follow directly from the displayed lists:

1. Two distinct supports in $\mathcal F$ meet in at most two coordinates.
2. A support in $\mathcal F$ and a support in $\mathcal H$ meet in at most two coordinates.
3. Two distinct supports in $\mathcal H$ meet in at most three coordinates.
4. If two distinct supports in $\mathcal H$ meet in three coordinates, then coordinate $8$ is one of the three, and the required signs on coordinate $8$ are opposite.

The last point can be checked compactly by writing each half-support as $\{8\}\cup T$ and coloring $T$ by the required sign at coordinate $8$. The negative triples are

$$\{\{0,2,5\},\{0,4,7\},\{0,9,10\},\{2,3,10\},\{2,4,9\},\{3,4,5\},\{3,7,9\},\{5,7,10\}\},$$

and the positive triples are

$$\{\{0,2,9\},\{0,4,5\},\{0,7,10\},\{2,3,4\},\{2,5,10\},\{3,5,7\},\{3,9,10\},\{4,7,9\}\}.$$

Within each color class, distinct triples intersect in at most one coordinate. Therefore two half-supports with the same sign at coordinate $8$ intersect in at most $1+1=2$ coordinates, including coordinate $8$. Thus any half-support intersection of size $3$ must involve opposite signs at coordinate $8$. In that case the two non-$8$ shared coordinates can contribute at most $+1+1$, while coordinate $8$ contributes $-1$, so the contribution is at most

$$1+1-1=1\le2.$$

For all other distinct support pairs, the largest possible dot product is at most the size of the support intersection, hence at most $2$. Finally, two distinct signings on the same four-support have dot product at most $2$; in the half-sign case the fixed coordinate $8$ contributes $+1$, and at least one of the other three signs differs. This proves the lemma.

### Lemma 2: Cavity-Cavity Compatibility

For distinct $u,v\in U$,

$$|\langle u,v\rangle|\le2.$$

For the centrally symmetric cavity $U\cup(-U)$, this absolute-value statement checks both $u,v$ and $u,-v$. The same-line antipodal pair $u,-u$ has inner product $-4$, so it is automatically valid.

Since the $f$-frame is orthonormal, all dot products are read directly from the block formulas. The following table gives the maximum value of $|\langle u,v\rangle|$ for each pair of blocks, excluding the identical line on the diagonal:

| $U_B$ |  | 2 | 0 | 0 | 0 | 0 | 2 | 2 | 2 | 2 | 0 |
| $U_C$ |  |  | 0 | 2 | 2 | 2 | 2 | 0 | 0 | 0 | 0 |
| $U_D$ |  |  |  | 2 | 2 | 2 | 2 | 2 | 0 | 0 | 2 |
| $U_E$ |  |  |  |  | 2 | 2 | 2 | 0 | 2 | 0 | 2 |

### Lemma 3: Core-Cavity Compatibility

For every $g\in G$ and every $u\in U\cup(-U)$,

$$\langle g,u\rangle\le2.$$

First, use the wall-load lemma. If all signings on a support $S$ are present, then for any row $x$,

$$\max_{\sigma_i=\pm1}\left\langle\sum_{i\in S}\sigma_i e_i,x\right\rangle=\sum_{i\in S}|x_i|.$$

If only some signings are present, the right-hand side remains an upper bound. For an axis row $\pm2e_i$, the condition is $|x_i|\le1$.

After converting the cavity blocks from the $f$-frame back to the standard basis, the following exact load certificate is obtained from the displayed support lists. Since $1+\sqrt2/2<2$, every entry is at most $2$.

| block | max $|x_i|$ on core axes | max $\sum_{i\in S}|x_i|$, $S\in\mathcal F$ | max $\sum_{i\in S}|x_i|$, $S\in\mathcal H$ |
|---|---:|---:|---:|
| $U_A$ | $1$ | $2$ | $1+\sqrt2/2$ |
| $U_B$ | $1$ | $2$ | $2$ |
| $U_C$ | $0$ | $2$ | $0$ |
| $U_D$ | $1$ | $2$ | $1+\sqrt2/2$ |
| $U_E$ | $1$ | $2$ | $1+\sqrt2/2$ |
| $U_F$ | $1$ | $2$ | $1+\sqrt2/2$ |
| $U_G$ | $0$ | $1+\sqrt2/2$ | $1+\sqrt2/2$ |
| $U_H$ | $1$ | $2$ | $2$ |
| $U_I$ | $1$ | $2$ | $2$ |
| $U_J$ | $1$ | $2$ | $2$ |
| $U_K$ | $0$ | $\sqrt2$ | $\sqrt2$ |

Thus every core-cavity inner product with $u\in U$ is at most $2$. Since the bounds use absolute values of the standard coordinates, the same load bounds apply to $-u$. Hence the table verifies both $G$ against $U$ and $G$ against $-U$.

### Proof Of The Theorem

By construction, every row in $G$, $U$, and $-U$ has squared norm $4$. Lemma 1 covers pairs inside $G$, Lemma 2 covers pairs inside $U\cup(-U)$, and Lemma 3 covers mixed pairs. Hence every distinct pair in $X$ has inner product at most $2$.

Since every vector in $X$ has norm $2$, the vectors themselves may be used as centers of unit spheres tangent to the unit sphere centered at the origin. For distinct $x,y\in X$,

$$
\|x-y\|^2=\|x\|^2+\|y\|^2-2\langle x,y\rangle
=8-2\langle x,y\rangle\ge4.
$$

Therefore the distance between any two outer centers is at least $2$, so the corresponding unit spheres do not overlap. This proves $K(11)\ge604$.

## Executable Appendix

The proof above is the mathematical certificate. The code cells below are included only as an executable appendix: they instantiate the listed supports and pair-frame formulas, check the finite dot-product and wall-load certificates, and parse the final single-line coordinate certificate. The verifier represents each coordinate exactly in $\mathbb{Q}(\sqrt2)$, as $(a+b\sqrt2)/2$ with rational $a,b$, and checks $|X|=604$, $\|x\|^2=4$ for every row, and $\langle x,y\rangle\le2$ for every distinct pair. No search, optimization, or preloaded coordinate data is used by the construction checker.


```python
#@title Appendix A: exact arithmetic and core builder
from __future__ import annotations

import hashlib
import itertools
import json
import math
import time
from collections import Counter
from fractions import Fraction

DIMENSION = 11

AXIS_COORDS = [0, 2, 3, 4, 5, 7, 9, 10]
FULL_SIGN_SUPPORTS = [
    (0, 1, 2, 10), (0, 1, 4, 9), (0, 1, 5, 7), (0, 2, 3, 7),
    (0, 2, 4, 6), (0, 3, 4, 10), (0, 3, 5, 9), (0, 5, 6, 10),
    (0, 6, 7, 9), (1, 2, 3, 9), (1, 2, 4, 5), (1, 3, 4, 7),
    (1, 3, 5, 10), (1, 7, 9, 10), (2, 3, 5, 6), (2, 4, 7, 10),
    (2, 5, 7, 9), (2, 6, 9, 10), (3, 4, 6, 9), (3, 6, 7, 10),
    (4, 5, 6, 7), (4, 5, 9, 10),
]
HALF_SIGN_SUPPORTS = [
    ((0, 2, 5, 8), 3, -1), ((0, 2, 8, 9), 2, 1),
    ((0, 4, 5, 8), 3, 1), ((0, 4, 7, 8), 3, -1),
    ((0, 7, 8, 10), 2, 1), ((0, 8, 9, 10), 1, -1),
    ((2, 3, 4, 8), 3, 1), ((2, 3, 8, 10), 2, -1),
    ((2, 4, 8, 9), 2, -1), ((2, 5, 8, 10), 2, 1),
    ((3, 4, 5, 8), 3, -1), ((3, 5, 7, 8), 3, 1),
    ((3, 7, 8, 9), 2, -1), ((3, 8, 9, 10), 1, 1),
    ((4, 7, 8, 9), 2, 1), ((5, 7, 8, 10), 2, -1),
]

HalfPair = tuple[int, int]
HalfRow = tuple[HalfPair, ...]


def construct_core() -> list[tuple[int, ...]]:
    rows: list[tuple[int, ...]] = []
    for i in AXIS_COORDS:
        for sign in (-1, 1):
            row = [0] * DIMENSION
            row[i] = 2 * sign
            rows.append(tuple(row))
    for support in FULL_SIGN_SUPPORTS:
        for signs in itertools.product((-1, 1), repeat=4):
            row = [0] * DIMENSION
            for idx, sign in zip(support, signs):
                row[idx] = sign
            rows.append(tuple(row))
    for support, local_position, required_sign in HALF_SIGN_SUPPORTS:
        for signs in itertools.product((-1, 1), repeat=4):
            if signs[local_position] != required_sign:
                continue
            row = [0] * DIMENSION
            for idx, sign in zip(support, signs):
                row[idx] = sign
            rows.append(tuple(row))
    assert len(rows) == 496
    return rows


def sign_zsqrt2(a: int, b: int) -> int:
    """Exact sign of a + b*sqrt(2)."""
    if a == 0 and b == 0:
        return 0
    if b == 0:
        return 1 if a > 0 else -1
    if a == 0:
        return 1 if b > 0 else -1
    if a > 0 and b > 0:
        return 1
    if a < 0 and b < 0:
        return -1
    left = a * a
    right = 2 * b * b
    if left == right:
        return 0
    if a > 0 and b < 0:
        return 1 if left > right else -1
    return 1 if right > left else -1


def abs_zsqrt2_pair(pair: HalfPair) -> HalfPair:
    a, b = pair
    return pair if sign_zsqrt2(a, b) >= 0 else (-a, -b)


def qsqrt2_le_integer(pair: HalfPair, rhs: int) -> bool:
    return sign_zsqrt2(pair[0] - rhs, pair[1]) <= 0


def add_pair(x: HalfPair, y: HalfPair) -> HalfPair:
    return (x[0] + y[0], x[1] + y[1])


def sub_pair(x: HalfPair, y: HalfPair) -> HalfPair:
    return (x[0] - y[0], x[1] - y[1])


def divide_by_sqrt2(pair: HalfPair) -> HalfPair:
    p, q = pair
    assert p % 2 == 0
    return (q, p // 2)


def integer_row_to_half_pairs(row: tuple[int, ...]) -> HalfRow:
    return tuple((2 * x, 0) for x in row)


def negate_half_row(row: HalfRow) -> HalfRow:
    return tuple((-p, -q) for p, q in row)


def canonical_line(row: HalfRow) -> HalfRow:
    return min(row, negate_half_row(row))


def half_row_norm_numerator(row: HalfRow) -> HalfPair:
    a = sum(p * p + 2 * q * q for p, q in row)
    b = sum(p * q for p, q in row)
    return (a, b)


def half_row_dot_numerator(left: HalfRow, right: HalfRow) -> HalfPair:
    a = 0
    b = 0
    for (p, q), (r, s) in zip(left, right):
        a += p * r + 2 * q * s
        b += p * s + q * r
    return (a, b)


def oriented_conflict_norm4(left: HalfRow, right: HalfRow) -> bool:
    # Dot product is (A+B*sqrt(2))/4. For norm-4 rows, normalized dot <= 1/2
    # is equivalent to A+B*sqrt(2) <= 8 when the dot is positive.
    a, b = half_row_dot_numerator(left, right)
    return sign_zsqrt2(a, b) > 0 and sign_zsqrt2(a - 8, b) > 0


def line_conflict_norm4(left: HalfRow, right: HalfRow) -> bool:
    a, b = half_row_dot_numerator(left, right)
    if sign_zsqrt2(a, b) < 0:
        a, b = -a, -b
    return sign_zsqrt2(a - 8, b) > 0


def qsqrt2_label(pair: HalfPair, denominator: int) -> str:
    p = Fraction(pair[0], denominator)
    q = Fraction(pair[1], denominator)
    if p == 0 and q == 0:
        return "0"
    pieces = []
    if p:
        pieces.append(str(p.numerator) if p.denominator == 1 else f"{p.numerator}/{p.denominator}")
    if q:
        sign = "-" if q < 0 else ("+" if pieces else "")
        aq = abs(q)
        if aq == 1:
            root = "sqrt(2)"
        elif aq.numerator == 1:
            root = f"sqrt(2)/{aq.denominator}"
        else:
            root = f"{aq.numerator}*sqrt(2)/{aq.denominator}"
        pieces.append(sign + root)
    return "".join(pieces)


def split_signed_terms(text: str) -> list[str]:
    s = text.strip().replace(" ", "").replace("sqrt(2)", "sqrt2")
    if not s or s == "0":
        return []
    s = s.replace("-", "+-")
    if s.startswith("+-"):
        s = s[1:]
    return [part for part in s.split("+") if part]


def parse_exact_coordinate(text: str) -> HalfPair:
    rational = Fraction(0)
    root = Fraction(0)
    for term in split_signed_terms(text):
        sign = -1 if term.startswith("-") else 1
        body = term[1:] if term.startswith("-") else term
        if "sqrt2" in body:
            tail = body.replace("sqrt2", "")
            coeff = Fraction(1)
            if tail.startswith("/"):
                coeff = Fraction(1, int(tail[1:]))
            elif tail.startswith("*"):
                coeff = Fraction(tail[1:])
            elif tail:
                raise ValueError(f"cannot parse term {term!r}")
            root += sign * coeff
        else:
            rational += sign * Fraction(body)
    p = 2 * rational
    q = 2 * root
    if p.denominator != 1 or q.denominator != 1:
        raise ValueError(f"coordinate is not a half-integer Q(sqrt2) value: {text!r}")
    return (int(p), int(q))


def parse_exact_rows(rows: list[list[str]]) -> list[HalfRow]:
    return [tuple(parse_exact_coordinate(value) for value in row) for row in rows]


def verify_norm4_kissing_configuration(rows: list[HalfRow]) -> dict[str, object]:
    bad_norms = []
    for idx, row in enumerate(rows):
        if half_row_norm_numerator(row) != (16, 0):
            bad_norms.append((idx, half_row_norm_numerator(row)))
    conflict_count = 0
    first_conflicts = []
    contact_count = 0
    inner_product_distribution: Counter[str] = Counter()
    max_dot = None
    max_pair = None
    for i in range(len(rows)):
        for j in range(i + 1, len(rows)):
            dot = half_row_dot_numerator(rows[i], rows[j])
            inner_product_distribution[qsqrt2_label(dot, 4)] += 1
            if max_dot is None or sign_zsqrt2(dot[0] - max_dot[0], dot[1] - max_dot[1]) > 0:
                max_dot = dot
                max_pair = (i, j)
            if sign_zsqrt2(dot[0] - 8, dot[1]) == 0:
                contact_count += 1
            if sign_zsqrt2(dot[0], dot[1]) > 0 and sign_zsqrt2(dot[0] - 8, dot[1]) > 0:
                conflict_count += 1
                if len(first_conflicts) < 5:
                    first_conflicts.append({"pair": [i, j], "dot_numerator_den4": dot})
    return {
        "valid": not bad_norms and conflict_count == 0,
        "num_spheres": len(rows),
        "dimension": DIMENSION,
        "bad_norm_count": len(bad_norms),
        "first_bad_norms": bad_norms[:5],
        "conflict_count": conflict_count,
        "first_conflicts": first_conflicts,
        "contact_count": contact_count,
        "inner_product_distribution": dict(inner_product_distribution),
        "max_normalized_inner_product": qsqrt2_label(max_dot or (0, 0), 16),
        "max_pair": list(max_pair) if max_pair is not None else None,
        "min_pair_distance": "2" if conflict_count == 0 and max_dot == (8, 0) else ">=2",
    }
```

```python
#@title Appendix B: explicit pair-frame cavity construction
CORE_WALL_CONSTRAINTS = (
    [((i,), 2) for i in AXIS_COORDS]
    + [(support, 4) for support in FULL_SIGN_SUPPORTS]
    + [(support, 4) for support, _local_position, _required_sign in HALF_SIGN_SUPPORTS]
)

NEG_TWO: HalfPair = (-4, 0)
NEG_ONE: HalfPair = (-2, 0)
ZERO: HalfPair = (0, 0)


def signed_one(sign: int) -> HalfPair:
    return (2 * sign, 0)


def signed_root2(sign: int) -> HalfPair:
    return (0, 2 * sign)


def frame_row(items: list[tuple[int, HalfPair]]) -> HalfRow:
    row: list[HalfPair] = [ZERO] * DIMENSION
    for axis, value in items:
        row[axis] = value
    return tuple(row)


def convert_cavity_frame_to_standard(frame: HalfRow) -> HalfRow:
    row = list(frame)
    d_coord = frame[6]
    a_coord = frame[8]
    row[6] = divide_by_sqrt2(add_pair(d_coord, a_coord))
    row[8] = divide_by_sqrt2(sub_pair(d_coord, a_coord))
    return tuple(row)


def construct_cavity_frame_blocks() -> dict[str, list[HalfRow]]:
    signs = (-1, 1)
    blocks: dict[str, list[HalfRow]] = {
        "A": [
            frame_row([(0, NEG_ONE), (1, signed_one(e)), (3, signed_one(h)), (6, signed_one(t))])
            for e, h, t in itertools.product(signs, repeat=3)
        ],
        "B": [
            frame_row([(0, NEG_ONE), (3, signed_one(e)), (8, signed_root2(h))])
            for e, h in itertools.product(signs, repeat=2)
        ],
        "C": [frame_row([(1, NEG_TWO)])],
        "D": [
            frame_row([(1, NEG_ONE), (2, signed_one(e)), (6, signed_one(h)), (7, signed_one(t))])
            for e, h, t in itertools.product(signs, repeat=3)
        ],
        "E": [
            frame_row([(1, NEG_ONE), (4, signed_one(e)), (6, signed_one(h)), (10, signed_one(t))])
            for e, h, t in itertools.product(signs, repeat=3)
        ],
        "F": [
            frame_row([(1, NEG_ONE), (5, signed_one(e)), (6, signed_one(h)), (9, signed_one(t))])
            for e, h, t in itertools.product(signs, repeat=3)
        ],
        "G": [
            frame_row([(1, NEG_ONE), (6, signed_one(e)), (8, signed_root2(h))])
            for e, h in itertools.product(signs, repeat=2)
        ],
        "H": [
            frame_row([(2, NEG_ONE), (7, signed_one(e)), (8, signed_root2(h))])
            for e, h in itertools.product(signs, repeat=2)
        ],
        "I": [
            frame_row([(4, NEG_ONE), (8, signed_root2(e)), (10, signed_one(h))])
            for e, h in itertools.product(signs, repeat=2)
        ],
        "J": [
            frame_row([(5, NEG_ONE), (8, signed_root2(e)), (9, signed_one(h))])
            for e, h in itertools.product(signs, repeat=2)
        ],
        "K": [frame_row([(6, NEG_TWO)])],
    }
    assert {name: len(rows) for name, rows in blocks.items()} == {
        "A": 8, "B": 4, "C": 1, "D": 8, "E": 8, "F": 8,
        "G": 4, "H": 4, "I": 4, "J": 4, "K": 1,
    }
    return blocks


def construct_explicit_cavity_lines_by_block() -> dict[str, list[HalfRow]]:
    blocks = construct_cavity_frame_blocks()
    standard_blocks: dict[str, list[HalfRow]] = {}
    seen: set[HalfRow] = set()
    for name, frame_rows in blocks.items():
        standard_rows = []
        for frame in frame_rows:
            assert half_row_norm_numerator(frame) == (16, 0)
            row = canonical_line(convert_cavity_frame_to_standard(frame))
            assert half_row_norm_numerator(row) == (16, 0)
            assert core_wall_filter(row)
            assert row not in seen
            seen.add(row)
            standard_rows.append(row)
        standard_blocks[name] = standard_rows
    assert len(seen) == 54
    return standard_blocks


def core_wall_load(row: HalfRow, support: tuple[int, ...]) -> HalfPair:
    total = (0, 0)
    for axis in support:
        total = add_pair(total, abs_zsqrt2_pair(row[axis]))
    return total


def core_wall_filter(row: HalfRow) -> bool:
    for support, rhs in CORE_WALL_CONSTRAINTS:
        if not qsqrt2_le_integer(core_wall_load(row, support), rhs):
            return False
    return True


def max_core_wall_load(rows: list[HalfRow]) -> HalfPair:
    best = (0, 0)
    for row in rows:
        for support, _rhs in CORE_WALL_CONSTRAINTS:
            load = core_wall_load(row, support)
            if sign_zsqrt2(load[0] - best[0], load[1] - best[1]) > 0:
                best = load
    return best


def max_cavity_frame_abs_dot(blocks: dict[str, list[HalfRow]]) -> HalfPair:
    rows = [row for block_rows in blocks.values() for row in block_rows]
    best = (0, 0)
    for i in range(len(rows)):
        for j in range(i + 1, len(rows)):
            dot = half_row_dot_numerator(rows[i], rows[j])
            if sign_zsqrt2(dot[0], dot[1]) < 0:
                dot = (-dot[0], -dot[1])
            if sign_zsqrt2(dot[0] - best[0], dot[1] - best[1]) > 0:
                best = dot
    return best


def flatten_blocks(blocks: dict[str, list[HalfRow]]) -> list[HalfRow]:
    rows: list[HalfRow] = []
    for name in ("A", "B", "C", "D", "E", "F", "G", "H", "I", "J", "K"):
        rows.extend(blocks[name])
    return rows


def expand_lines_to_oriented_rows(lines: list[HalfRow]) -> list[HalfRow]:
    rows: list[HalfRow] = []
    for row in lines:
        rows.append(row)
        rows.append(negate_half_row(row))
    return rows
```

```python
#@title Appendix C: instantiate and verify the explicit construction
start_time = time.time()
core_rows = construct_core()
core_half_rows = [integer_row_to_half_pairs(row) for row in core_rows]

frame_blocks = construct_cavity_frame_blocks()
cavity_blocks = construct_explicit_cavity_lines_by_block()
cavity_lines = flatten_blocks(cavity_blocks)
constructed_cavity = expand_lines_to_oriented_rows(cavity_lines)
constructed_rows = core_half_rows + constructed_cavity
constructed_metrics = verify_norm4_kissing_configuration(constructed_rows)

core_set = set(core_half_rows)
cavity_set = set(constructed_cavity)
core_cavity_overlap = core_set & cavity_set
distinct_row_count = len(set(constructed_rows))

max_frame_abs_dot = max_cavity_frame_abs_dot(frame_blocks)
max_wall_load = max_core_wall_load(cavity_lines)

assert len(core_rows) == 496
assert len(cavity_lines) == 54
assert len(constructed_cavity) == 108
assert len(core_cavity_overlap) == 0
assert distinct_row_count == 604
assert qsqrt2_le_integer(max_frame_abs_dot, 8)  # actual dot <= 2, numerator denominator is 4
assert qsqrt2_le_integer(max_wall_load, 4)      # actual wall load <= 2, numerator denominator is 2
assert constructed_metrics["num_spheres"] == 604
assert constructed_metrics["conflict_count"] == 0
assert constructed_metrics["valid"] is True

print(json.dumps({
    "core_vectors": len(core_rows),
    "cavity_lines": len(cavity_lines),
    "cavity_oriented_rows": len(constructed_cavity),
    "distinct_rows": distinct_row_count,
    "core_cavity_overlap": len(core_cavity_overlap),
    "cavity_block_counts": {name: len(rows) for name, rows in cavity_blocks.items()},
    "max_cavity_line_abs_dot": qsqrt2_label(max_frame_abs_dot, 4),
    "max_core_wall_load": qsqrt2_label(max_wall_load, 2),
    "constructed_metrics": constructed_metrics,
    "elapsed_seconds": round(time.time() - start_time, 3),
}, indent=2, sort_keys=True))
```
